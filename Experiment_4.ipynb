```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

X = np.array([1.1, 1.3, 1.5, 2.0, 2.2, 2.9, 3.0, 3.2, 3.7, 4.0, 2.9])
Y = np.array([39000, 46000, 37700, np.nan, 39800,
              56600, 60100, 54400, 57100, 63200, 56600])

df = pd.DataFrame({"Experience": X, "Salary": Y})
print(df)

print("\nMissing Values:")
print(df.isnull().sum())
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# Clean the data
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df = df.drop_duplicates()
print(df.isnull().sum())
print(df.duplicated().sum())

# Split
X = df[["Experience"]]
Y = df["Salary"]
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=1)
print(X_train.shape, X_test.shape)

# Train and predict
model = LinearRegression()
model.fit(X_train, Y_train)
Y_pred = model.predict(X_test)
print("prediction value", Y_pred)
print("R2 Score:", r2_score(Y_test, Y_pred))

# Predict for 3 years (the print text must match the number used)
experience = np.array([[3]])
salary = model.predict(experience)
print("Predicted Salary for 3 years of Experience:", salary[0])

# Plot
plt.plot(X, Y, label="Actual Salary")
plt.plot(X, model.predict(X), label="Regression Line")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Linear Regression - Experience vs Salary")
plt.legend()
plt.grid()
plt.show()

print(df.mean())
print(df.median())
print(df.std())
```
