```python


# Step 1: Import libraries
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Step 2: Load the dataset (your file if uploaded, else practice data)
try:
    data = pd.read_csv("/content/Pass-Fail Data.csv")
    print("Using uploaded file: Pass-Fail Data.csv")
except FileNotFoundError:
    print("File not found. Using PRACTICE data, not the lab dataset.")
    rng = np.random.default_rng(42)
    n = 100
    data = pd.DataFrame({
        "student_id": range(1, n + 1),
        "attendance_pct": rng.integers(35, 100, n),
        "homework_pct": rng.integers(35, 100, n),
        "midterm_score": rng.integers(35, 100, n),
        "study_hours_per_week": rng.integers(1, 15, n),
    })
    score = (0.3 * data["attendance_pct"] + 0.3 * data["homework_pct"]
             + 0.3 * data["midterm_score"] + 1.5 * data["study_hours_per_week"]
             + rng.normal(0, 5, n))
    data["pass"] = (score > score.median()).astype(int)

# Step 3: Display the dataset
print("First 5 rows of dataset:")
print(data.head())

# Step 4: Dataset information
print("\nDataset Information:")
print(data.info())

# Step 5: Column names
print("\nColumn Names:")
print(data.columns)

# Step 6: Missing values
print("\nMissing Values:")
print(data.isnull().sum())

# Step 7: Duplicate values
print("\nDuplicate Values:")
print(data.duplicated().sum())

# Step 8: Input and output variables
X = data[["attendance_pct", "homework_pct", "midterm_score", "study_hours_per_week"]]
y = data["pass"]

# Step 9: Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
print("\nTraining data size:", X_train.shape)
print("Testing data size:", X_test.shape)

# Step 10: Create and train the model
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

# Step 11: Predict
y_pred = model.predict(X_test)
print("\nActual Values:")
print(y_test.values)
print("\nPredicted Values:")
print(y_pred)

# Step 12: Accuracy
accuracy = accuracy_score(y_test, y_pred)
print("\nAccuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100, "%")

# Step 13: Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# Step 14: Classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Step 15: Predict a new student
new_student = pd.DataFrame(
    [[85, 80, 75, 10]], columns=X.columns
)
prediction = model.predict(new_student)
print("\nNew Student Prediction:")
if prediction[0] == 1:
    print("Student will PASS")
else:
    print("Student will FAIL")

# Step 16: Prediction probability
probability = model.predict_proba(new_student)
print("\nPrediction Probability:")
print("Fail Probability:", probability[0][0])
print("Pass Probability:", probability[0][1])

# Step 17: Model coefficients
print("\nModel Coefficients:")
for column, coefficient in zip(X.columns, model.coef_[0]):
    print(column, ":", coefficient)
print("\nIntercept:", model.intercept_[0])
```
