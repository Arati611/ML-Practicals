```python
# Create a practice Country-data.csv
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)

# Three groups of countries: developed, developing, low-income
sizes = [36, 84, 47]
groups = {
    # child_mort, exports, health, imports, income, inflation, life_expec, total_fer, gdpp
    "developed":  dict(m=[5, 45, 9, 42, 38000, 3, 79, 1.7, 35000],
                       s=[2, 20, 2, 18, 12000, 2, 2, 0.3, 14000]),
    "developing": dict(m=[22, 38, 6, 45, 12000, 8, 72, 2.5, 6000],
                       s=[10, 16, 2, 16, 6000, 6, 4, 0.8, 4000]),
    "low_income": dict(m=[85, 30, 6, 40, 3000, 12, 60, 5.0, 1200],
                       s=[30, 14, 2.5, 15, 2000, 9, 5, 1.0, 1000]),
}
cols = ["child_mort", "exports", "health", "imports", "income",
        "inflation", "life_expec", "total_fer", "gdpp"]

rows = []
for (name, g), n in zip(groups.items(), sizes):
    rows.append(rng.normal(g["m"], g["s"], size=(n, 9)))
data = np.vstack(rows)

df = pd.DataFrame(data, columns=cols)

# Keep values realistic (no negatives, sensible limits)
df["child_mort"] = df["child_mort"].clip(2, 200).round(1)
df["exports"] = df["exports"].clip(5, 120).round(1)
df["health"] = df["health"].clip(2, 15).round(2)
df["imports"] = df["imports"].clip(5, 120).round(1)
df["income"] = df["income"].clip(500, 120000).round(0).astype(int)
df["inflation"] = df["inflation"].clip(-2, 50).round(2)
df["life_expec"] = df["life_expec"].clip(40, 85).round(1)
df["total_fer"] = df["total_fer"].clip(1, 8).round(2)
df["gdpp"] = df["gdpp"].clip(300, 100000).round(0).astype(int)

# Shuffle the rows and add practice country names
df = df.sample(frac=1, random_state=42).reset_index(drop=True)
df.insert(0, "country", [f"Country_{i:03d}" for i in range(1, len(df) + 1)])

df.to_csv("/content/Country-data.csv", index=False)
print(df.shape)
print(df.head())

# Experiment 9: Country Clustering using K-Means

import glob
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Step 1: Load the data
files = glob.glob("/content/*ountry*.csv")
df = pd.read_csv(files[0])
print("Using file:", files[0])
print("Shape:", df.shape)
print("\nFirst 5 Records:")
print(df.head())

# Step 2: Remove the country name (text) to get the inputs
x = df.drop('country', axis=1)

# Step 3: Scale the data so all columns have a similar range
scaler = StandardScaler()
x_scaled = pd.DataFrame(scaler.fit_transform(x), columns=x.columns)
print("\nScaled data:")
print(x_scaled.head())

# Step 4: Create and train K-Means with 3 clusters
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans.fit(x_scaled)

# Step 5: Add the cluster number to the table
df['cluster'] = kmeans.labels_
print("\nCountries with Cluster:")
print(df[['country', 'cluster']].head(15))

# Step 6: Count countries in each cluster
print("\nNumber of countries in each cluster:")
print(df['cluster'].value_counts().sort_index())

# Step 7: Predict the cluster of a new country
n
```
